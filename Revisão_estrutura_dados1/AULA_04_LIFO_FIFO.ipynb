Exercicio 1: O simulador desfazer:

```python
from collections import deque


class FilaAcoes:
    def __init__(self):
        self.fila = deque()
        self.historico = []

    def adicionar(self, texto):
        self.fila.append(texto)
        print(f"'{texto}' adicionado à fila.")

    def apagar(self):
        if len(self.fila) > 0:
            texto_removido = self.fila.pop()
            self.historico.append(texto_removido)
            print(f"'{texto_removido}' foi apagado.")
        else:
            print("A fila está vazia.")

    def desfazer(self):
        if len(self.historico) > 0:
            texto = self.historico.pop()
            self.fila.append(texto)
            print(f"'{texto}' foi restaurado.")
        else:
            print("Não há nenhuma ação para desfazer.")

    def mostrar(self):
        if len(self.fila) == 0:
            print("A fila está vazia.")
        else:
            print("\n--- FILA DE NOMES ---")
            for texto in self.fila:
                print(texto)


Fila = FilaAcoes()


while True:
    print("\n--- EDITOR DE TEXTO ---")
    print("Digite 1: Digitar nome.")
    print("Digite 2: Apagar.")
    print("Digite 3: Desfazer.")
    print("Digite 4: Mostrar fila.")
    print("Digite 5: Sair.")

    opcao = int(input("\nDigite a opção: "))

    if opcao == 1:
        texto = input("Digite o nome: ")
        Fila.adicionar(texto)

    elif opcao == 2:
        Fila.apagar()

    elif opcao == 3:
        Fila.desfazer()

    elif opcao == 4:
        Fila.mostrar()

    elif opcao == 5:
        print("Encerrando...")
        break

    else:
        print("Opção inválida!")

```
