Exercicio 1: O simulador desfazer:

```python
from collections import deque


class FilaAcoes:
  def __init__(self):
    self.fila = deque()

  def adicionar(self, texto):
    self.fila.append(texto)
    print(f"'{texto}' adicionada a fila.")
    pass

Fila = FilaAcoes()

while True:
  print("-EDITOR DE TEXTO-")
  print("Digite 1: Digitar nome.")
  print("Digite 2: apagar.")
  print("Digite 3: desfazer.")
  print("Digite 4: mostrar fila.")
  print("Digite 5: sair.")

  opcao = int(input("Digite a opção: "))

  if opcao == 1:
    texto = input("Digite o texto: ")
    Fila.adicionar(texto)

  elif opcao == 2:
    Fila.apagar()

  elif opcao == 3:
    Fila.desfazer()

  elif opcao == 4:
    Fila.substituir()

  elif opcao == 5:
    print("Encerrando...")
    break

  else:
    print("Opção inválida")


```
