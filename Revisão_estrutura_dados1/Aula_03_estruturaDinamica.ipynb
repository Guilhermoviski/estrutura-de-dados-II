Exercício 2 — Construindo uma cadeia

Crie:

```python
n1 = Node("A")
n2 = Node("B")
n3 = Node("C")
```

Conecte:

```text
A → B → C → None
```

Depois percorra a estrutura e produza:

```text
A
B
C
```

````python
class Node:
    def __init__(self, valor):
        self.valor = valor
        self.proximo = None

n1  = Node("A")
n2  = Node("B")
n3  = Node("C")

n1.proximo = n2
n2.proximo = n3

atual = n1
while atual:
    print(atual.valor)
    atual = atual.proximo
````

Desafio do Atendimento em Python

````python
class Paciente:
    def __init__(self, nome, idade, prioridade):
        self.nome = nome
        self.idade = idade
        self.prioridade = prioridade


class Node:
    def __init__(self, paciente):
        self.paciente = paciente
        self.proximo = None


class FilaAtendimento:
    def __init__(self):
        self.inicio = None
        self.fim = None
        self._tamanho = 0

    def esta_vazia(self):
        return self.inicio is None

    def tamanho(self):
        return self._tamanho

    def adicionar(self, paciente):
        novo = Node(paciente)

        # Fila vazia
        if self.esta_vazia():
            self.inicio = novo
            self.fim = novo

        # Paciente prioritário
        elif paciente.prioridade == "Prioridade":

            # Se o primeiro for normal, entra no início
            if self.inicio.paciente.prioridade == "Normal":
                novo.proximo = self.inicio
                self.inicio = novo

            else:
                # Procura o último paciente prioritário
                atual = self.inicio

                while (
                    atual.proximo is not None
                    and atual.proximo.paciente.prioridade == "Prioridade"
                ):
                    atual = atual.proximo

                novo.proximo = atual.proximo
                atual.proximo = novo

                if novo.proximo is None:
                    self.fim = novo

        # Paciente normal vai para o final
        else:
            self.fim.proximo = novo
            self.fim = novo

        self._tamanho += 1

        print(f"\nPaciente {paciente.nome} adicionado com sucesso!")

    def atender(self):
        if self.esta_vazia():
            print("\nA fila está vazia!")
            return None

        paciente_atendido = self.inicio.paciente

        self.inicio = self.inicio.proximo
        self._tamanho -= 1

        # Se removeu o último paciente
        if self.inicio is None:
            self.fim = None

        return paciente_atendido

    def listar(self):
        if self.esta_vazia():
            print("\nA fila está vazia!")
            return

        atual = self.inicio
        posicao = 1

        print("\n--- FILA DE ATENDIMENTO ---")

        while atual is not None:
            paciente = atual.paciente

            print(
                f"{posicao}. {paciente.nome} - "
                f"{paciente.idade} anos - "
                f"{paciente.prioridade}"
            )

            atual = atual.proximo
            posicao += 1

        print(f"\nTotal de pacientes: {self.tamanho()}")


# =========================
# MENU PRINCIPAL
# =========================

fila = FilaAtendimento()

while True:
    print("\n" + "=" * 30)
    print(" SISTEMA DE ATENDIMENTO")
    print("=" * 30)
    print("1 - Adicionar paciente")
    print("2 - Atender paciente")
    print("3 - Mostrar fila")
    print("4 - Mostrar tamanho da fila")
    print("5 - Verificar se a fila está vazia")
    print("0 - Sair")

    opcao = input("\nEscolha uma opção: ")

    # OPÇÃO 1 - ADICIONAR
    if opcao == "1":
        nome = input("Nome do paciente: ")

        try:
            idade = int(input("Idade: "))
        except ValueError:
            print("Idade inválida!")
            continue

        print("\nTipo de atendimento:")
        print("1 - Normal")
        print("2 - Prioridade")

        tipo = input("Escolha: ")

        if tipo == "1":
            prioridade = "Normal"
        elif tipo == "2":
            prioridade = "Prioridade"
        else:
            print("Opção inválida!")
            continue

        paciente = Paciente(nome, idade, prioridade)
        fila.adicionar(paciente)

    # OPÇÃO 2 - ATENDER
    elif opcao == "2":
        paciente = fila.atender()

        if paciente:
            print(f"\nAtendendo: {paciente.nome}")
            print(f"Idade: {paciente.idade}")
            print(f"Tipo: {paciente.prioridade}")

    # OPÇÃO 3 - LISTAR FILA
    elif opcao == "3":
        fila.listar()

    # OPÇÃO 4 - TAMANHO DA FILA
    elif opcao == "4":
        print(f"\nQuantidade de pacientes na fila: {fila.tamanho()}")

    # OPÇÃO 5 - FILA VAZIA
    elif opcao == "5":
        if fila.esta_vazia():
            print("\nA fila está vazia!")
        else:
            print("\nExistem pacientes aguardando atendimento.")

    # OPÇÃO 0 - SAIR
    elif opcao == "0":
        print("\nSistema encerrado!")
        break

    # OPÇÃO INVÁLIDA
    else:
        print("\nOpção inválida! Tente novamente.")
````
