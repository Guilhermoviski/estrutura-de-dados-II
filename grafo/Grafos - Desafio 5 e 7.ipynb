Desafio 5 - Isomorfismo

Dois grafos podem ter desenhos e rótulos diferentes e ainda apresentar a mesma estrutura. Compare os grafos a seguir:

G: V = {1, 2, 3, 4}; E = {(1,2), (2,3), (3,4), (4,1)}

H: V = {a, b, c, d}; E = {(a,c), (c,b), (b,d), (d,a)}

O que fazer

1.	Determine a ordem e o tamanho de cada grafo.

2.	Calcule o grau de cada vértice e escreva a sequência de graus de G e H.

3.	Procure uma correspondência entre os vértices de G e os de H.

4.	Verifique, aresta por aresta, se a correspondência preserva as adjacências.

5.	Conclua se os grafos são isomorfos e justifique.

Respostas:

vértices: {1,2,3,4}

Arestas: {(1,2), (2,3), (3,4), (4,1)}

Ordem: 4 (quantidades de vértices)

Tamanho: 4 (quantidade de arestas)

Grafo H:

vértices: {a, b, c, d}

Arestas: {(a, b), (b, c), (c, d), (d,a)}

Ordem: 4 (quantidades de vértices)

Tamanho: 4 (quantidade de arestas)

Grafo H:

| Vértice | Grau |

arestas grafo h: {(a, b), (b, c), (c, d), (d,a)}

preserva todas as adjacências:

(1,2) = (a, b)

Justificativa: Os grafos G e H são isomorfos, pois possuem a mesma quantidade de vértices e arestas, a mesma sequência de graus e uma correspondência entre os vértices que preserva todas as adjacências. Portanto, apresentam a mesma estrutura.

Reflexão: ter a mesma sequência de graus é suficiente para garantir isomorfismo? Explique.

Não. Ter a mesma sequência de graus não garante que dois grafos sejam isomorfos, pois eles podem apresentar conexões diferentes entre os vértices.
