Exercício 2 — Construindo uma cadeia

Crie:

```python
n1 = Node("A")
n2 = Node("B")
n3 = Node("C")
```

Conecte:

```text
A → B → C → None
```

Depois percorra a estrutura e produza:

```text
A
B
C
```

````python
class Node:
    def __init__(self, valor):
        self.valor = valor
        self.proximo = None

n1  = Node("A")
n2  = Node("B")
n3  = Node("C")

n1.proximo = n2
n2.proximo = n3

atual = n1
while atual:
    print(atual.valor)
    atual = atual.proximo
````

Exercício 3 — Depuração

Analise:

```python
n1 = Node(10)
n2 = Node(20)
n3 = Node(30)

n1.proximo = n2
n2.proximo = n3

atual = n1

while atual is not None:
    print(atual.valor)
```

Perguntas:

1. Qual é o problema?
2. Por que o programa não termina?
3. Qual linha deve ser acrescentada?
4. Qual será a saída depois da correção?

Respostas:
1. o problema está no while, ele não vai para o proximo valor.
2.  como não tem o codigo para mudar o atual.valor, ele sempre continua no n1, nisso cria um loop.
3. no while deveria ser acrescentado a linha para avançar a atual linha:
atual = atual.proximo
4. a saída depois da correção é: 10, 20, 30.

````python
n1 = Node(10)
n2 = Node(20)
n3 = Node(30)

n1.proximo = n2
n2.proximo = n3

atual = n1

while atual is not None:
    print(atual.valor)
    atual = atual.proximo
````
